```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

## Linear Regression

```python
x = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([2, 4, 5, 4, 5])
```

```python
model = LinearRegression()
model.fit(x, y)
```

```python
y_pred = model.predict(x)
```

```python
print(f"Slope: {model.coef_[0]}")
print(f"Intercept: {model.intercept_}")
```

```python
n = len(y)
mse = np.sum((y - y_pred) ** 2) / n
rmse = np.sqrt(mse)
```

```python
print(f"MSE: {mse}")
print(f"RMSE: {rmse}")
```

```python
plt.scatter(x, y, color='blue', label='Actual')
plt.plot(x, y_pred, color='red', label='Predicted')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.show()
```

## Multivariable Regression

```python
X1 = np.array([0, 5, 15, 25, 35, 45, 55, 60])
X2 = np.array([1, 1, 2, 5, 11, 15, 34, 35])
y = np.array([4, 5, 20, 14, 32, 22, 38, 43])
X = np.column_stack((X1, X2))
```

```python
print("Manual Method:")
n = len(y)
X_with_intercept = np.column_stack((np.ones(n), X))
beta = np.linalg.inv(X_with_intercept.T @ X_with_intercept) @ X_with_intercept.T @ y
y_pred_manual = X_with_intercept @ beta
```

```python
print(f"Intercept: {beta[0]}")
print(f"Coefficient X1: {beta[1]}")
print(f"Coefficient X2: {beta[2]}")
```

```python
mse_manual = np.sum((y - y_pred_manual) ** 2) / n
rmse_manual = np.sqrt(mse_manual)
print(f"MSE (Manual): {mse_manual}")
print(f"RMSE (Manual): {rmse_manual}")
```

```python
print("\nLibrary Method:")
model = LinearRegression()
model.fit(X, y)
y_pred_lib = model.predict(X)
```

```python
print(f"Intercept: {model.intercept_}")
print(f"Coefficient X1: {model.coef_[0]}")
print(f"Coefficient X2: {model.coef_[1]}")
```

```python
mse_lib = np.sum((y - y_pred_lib) ** 2) / n
rmse_lib = np.sqrt(mse_lib)
print(f"MSE (Library): {mse_lib}")
print(f"RMSE (Library): {rmse_lib}")
```

```python
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(X1, X2, y, color='blue', marker='o', s=100, label='Actual')
ax.scatter(X1, X2, y_pred_lib, color='red', marker='^', s=100, label='Predicted')
ax.set_xlabel('X1')
ax.set_ylabel('X2')
ax.set_zlabel('Y')
ax.legend()
plt.show()
```

## SAT vs GPA Linear Regression

```python
data = pd.read_csv('simpleregdataset.csv')
X_sat = data['SAT'].values.reshape(-1, 1)
y_gpa = data['GPA'].values
```

```python
model_sat = LinearRegression()
model_sat.fit(X_sat, y_gpa)
```

```python
y_gpa_pred = model_sat.predict(X_sat)
```

```python
print(f"Slope: {model_sat.coef_[0]}")
print(f"Intercept: {model_sat.intercept_}")
```

```python
n_sat = len(y_gpa)
mse_sat = np.sum((y_gpa - y_gpa_pred) ** 2) / n_sat
rmse_sat = np.sqrt(mse_sat)
```

```python
print(f"MSE: {mse_sat}")
print(f"RMSE: {rmse_sat}")
```

```python
plt.scatter(X_sat, y_gpa, color='blue', label='Actual', alpha=0.6)
plt.plot(X_sat, y_gpa_pred, color='red', label='Predicted', linewidth=2)
plt.xlabel('SAT Score')
plt.ylabel('GPA')
plt.legend()
plt.show()
```
